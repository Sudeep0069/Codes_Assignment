```python
def oddeven(n):
  # This function checks whether a number is odd or even
  return "Even" if n%2==0 else "Odd"
```

```python
def check(n):
# Check if a number is positive, negative or zero
  if n>0:
    print(n,"is positive")
  else:
    print(f"{n} is","negative" if n<0 else "zero")
```

```python
def larger(a,b):
# Returns the larger of two numbers
  return a if a>b else b
```

```python
a,b=map(int,input("enter 2 numbers separated by ',': ").split(','))
print("Larger of the two: ",larger(a,b))
```

```python
def larger3(a,b,c):
  if a>b and a>c:
    return a
  elif b>a and b>c:
    return b
  else:
    return c
```

```python
a,b,c=map(int,input("enter 3 numbers separated by ',': ").split(','))
print("Larger of the three: ",larger3(a,b,c))
```

```python
def factorial(n):
  if n == 0 or n == 1:
    return 1
  i=1
  for _ in range(1,n+1):
    i*=_
  return i
```

```python
factorial(10)
```

```python
def count_digits(n):
  if n==0:
    return 1
  c=0
  while n>0:
    n=n//10
    c+=1
  return c
```

```python
count_digits(100)
```

```python
def reverse_number(n):
  rev=0
  while n>0:
    rev=rev*10+n%10
    n=n//10
  return rev
```

```python
reverse_number(1918)
```

```python
def check_prime(n):
  if n<=1:
    return "Not Prime. Number must be > 1"
  for i in range(2,n//2+1):
    if n%i==0:
      return "Not prime"
  return "Prime"
```

```python
check_prime(79)
```

```python
def count_characters(text):
  c=0
  for i in text:
    c+=1
  return c
```

```python
def count_vowels(text):
  c=0
  for i in text:
    if i in ['a','e','i','o','u']:
      c+=1
  return c
```

```python
count_vowels('education')
```

```python
def count_consonants(text):
  c=0
  for i in text:
    if i not in ['a','e','i','o','u',' ']:
      c+=1
  return c
```

```python
count_consonants('where are you')
```

```python
def count_vowels_consonants(text):
  v=c=0
  for i in text:
    if i in ['a','e','i','o','u']:
      v+=1
    elif i!=' ':
      c+=1
  return v,c
```

```python
v,c=count_vowels_consonants('Python Programming ')
print("vowels:",v,"\nconsonants:",c)
```

```python
def count_words(text):
  text = text.strip()
  if text == '':
    return 0 #return 0 if there are no words
  if ' ' in text:
    c=1
    for i in range(len(text)):
      if text[i] == ' ' and text[i+1]!= ' ':
        c+=1 #count one word per space
    return c
  else:
    return 1 #if there are no spaces and str is not empty, it means therer's only 1 word
```

```python
count_words("  Python is     easy to learn")
```

```python
def character_frequency(text, ch):
  c=0
  for i in text:
    if i == ch:
     c+=1
  return c
```

```python
character_frequency('programming','g')
```

```python
def convert_uppercase2(text):
  s=''
  for i in text:
    if 'a'<= i <= 'z':
      s+= chr(ord(i) - 32) # ord() gives the ordinal number of i & chr() converts the new ordinal number into chr
    else:
      s+=i
  return s
```

```python
convert_uppercase2('Hi, i am jarvis!')
```

```python
def count_case(text):
  u=l=d=s=0
  for i in text:
    if i.isupper():
      u+=1
    elif i.islower():
      l+=1
    elif i.isdigit():
      d+=1
    elif i.isspace():
      s+=1
  return u,l,d,s
```

```python
a,b,c,d=count_case('Python123 ABC')
print(f'uppercase: {a}\nlowercase: {b}\ndigit: {c}\nspace: {d}')
```

```python
def find_longest_word(text):
  w= text.replace(",", "").replace("?", "").replace(".","").replace("!","").split(' ')
  for i in w:
    fl=1
    for j in w:
      if len(i)<len(j): # compare lengths. if < any other word's length, flag it
        fl = 0
        break
    if fl == 1:
      return i
```

```python
find_longest_word('hey babe, wassup? how you doin?!')
```

```python
def vowel_occurance(text):
  text=text.lower()
  a=e=i=o=u=0
  for j in text:
    if j == 'a':
      a+=1
    elif j == 'e':
      e+=1
    elif j == 'i':
      i+=1
    elif j == 'o':
      o+=1
    elif j == 'u':
      u+=1
  return a,e,i,o,u
```

```python
a,e,i,o,u = vowel_occurance('Education is powEr!')
print(f'A: {a}\nE: {e}\nI: {i}\nO: {o}\nU: {u}')
```

# Section D – Loops, Patterns and Menus

```python
def star_pattern(n): # using nested loops
  for i in range(1,n+1):
    for j in range(1,i+1):
      print('*',end='')
    print()
```

```python
star_pattern(5)
```

```python
def star_pyramid(n): # all in one go
  for i in range(1,n+1):
    print(' '*(n-i)+'*'*i+'*'*(i-1))
```

```python
star_pyramid(7)
```

```python
def number_pattern(n):
  for i in range(1,n+1):
    for j in range(1,i+1):
      print(j,end='')
    print()
```

```python
number_pattern(5)
```
